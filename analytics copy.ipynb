Importera paket och importera csv-filen med data till en DataFrame

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv('health_study_dataset.csv')
```

Räkna ut medel, median, min och max för: age, weight, height, systolic_bp, cholesterol.

```python
# Avarage age
print(f'Avarge age: {df['age'].mean():.1f} years')

# Avarage weight
print(f'Avarge weight: {df['weight'].mean():.1f} Kg')

# Avarage height
print(f'Avarage height: {df['height'].mean():.1f} cm')

# Avarage systolic_bp
print(f'Avarage systolic bp: {df['systolic_bp'].mean():.1f} mmHg')

# Avarage  cholesterol
print(f'Avarage systolic bp: {df['cholesterol'].mean():.1f} mmol/L \n')


print('*' * 50)


# Median age
print(f'\nMedian age: {df['age'].median():.1f} years')

# Median weight
print(f'Median weight: {df['weight'].median():.1f} Kg')

# Median height
print(f'Median height: {df['height'].median():.1f} cm')

# Median systolic_bp
print(f'Median systolic bp: {df['systolic_bp'].median():.1f} mmHg')

# Median  cholesterol
print(f'Median systolic bp: {df['cholesterol'].median():.1f} mmol/L \n')


print('*' * 50)


# Min age
print(f'\nMin age: {df['age'].min():.1f} years')

# Min weight
print(f'Min weight: {df['weight'].min():.1f} Kg')

# Min height
print(f'Min height: {df['height'].min():.1f} cm')

# Min systolic_bp
print(f'Min systolic bp: {df['systolic_bp'].min():.1f} mmHg')

# Min  cholesterol
print(f'Min systolic bp: {df['cholesterol'].min():.1f} mmol/L \n')


print('*' * 50)


# Max age
print(f'\nMax age: {df['age'].max():.1f} years')

# Max weight
print(f'Max weight: {df['weight'].max():.1f} Kg')

# Max height
print(f'Max height: {df['height'].max():.1f} cm')

# Max systolic_bp
print(f'Max systolic bp: {df['systolic_bp'].max():.1f} mmHg')

# Max  cholesterol
print(f'Max systolic bp: {df['cholesterol'].max():.1f} mmol/n')
```

```python
cols1 = ['age', 'weight', 'height', 'systolic_bp', 'cholesterol']
cols2 = ['mean', 'median', 'min', 'max']

for a in cols1:
    for b in cols2:
        print(f'{b} {a}: {df[a].agg(b):.1f}')
    print('*'*50)
```

```python
cols = ['age', 'weight', 'height', 'systolic_bp', 'cholesterol']
funcs = ['mean', 'median', 'min', 'max']
unit = ['years', 'Kg', 'cm', 'mmHg', 'mmol']

for col in range(len(cols)):
    for func in range(len(funcs)):
        print(f'{funcs[func]} {cols[col]}: {df[cols[col]].agg(funcs[func]):.1f} {unit[col]}')
    print('*'*50)
```

```python
cols = {'age': 'Age', 'weight': 'Weight', 'height': 'Height', 'systolic_bp': 'Systolic_bp', 'cholesterol': 'Cholesterol'}
funcs = {'mean': 'Avarage', 'median': 'Median', 'min': 'Min', 'max':'Max'}
unit = ['years', 'Kg', 'cm', 'mmHg', 'mmol']
counter = 0

for col_key, col_value in cols.items():
    for func_key, func_value in funcs.items():
        print(f'{func_value} {col_value}: {df[col_key].agg(func_key):.1f} {unit[counter]}')
    counter += 1
    print('')
    print('*'*50)
```

```python
cols = ['age', 'weight', 'height', 'systolic_bp', 'cholesterol']
funcs = ['mean', 'median', 'min', 'max']
units = {
    'age': 'years',
    'weight': 'Kg',
    'height': 'cm',
    'systolic_bp': 'mmHg',
    'cholesterol': 'mmol',
}

df_cols = df[cols].agg(funcs).T.round(1)

df_cols['units'] = df_cols.index.map(units)

df_cols
```

```python
cols = ['age', 'weight', 'height', 'systolic_bp', 'cholesterol']
funcs = ['mean', 'median', 'min', 'max']

df_cols = df[cols].agg(funcs).round(1)


units = {
    'age': 'years',
    'weight': 'Kg',
    'height': 'cm',
    'systolic_bp': 'mmHg',
    'cholesterol': 'mmol',
}

units_row = pd.DataFrame([units], index=['units'])
df_cols = pd.concat([df_cols, units_row])


df_cols
```
